## Introduction

Deep learning libraries like Torch utilize autodifferentiation of tensors to compute the parameter updates necessary to learn complex models from data. This technique is central to understanding why deep learning has become so widely used and effective. The autodifferentiation process is a neat trick that builds up a computational graph and then uses that graph to provide derivatives for user-constructed mathematical functions. At heart, this process is just an instantiation of the chain-rule based on implementations of every function and its derivative.

However, a library still *needs* to have efficient implementations of derivatives for its key building blocks. This sounds trivial -> just implement high school calculus. However, this is a bit more tricky than it sounds. Tensor-to-tensor functions are pretty complex and require keeping careful track of indexing on both the input and the output side.

Your goal in these puzzles is to implement the Jacobian, i.e. the derivative of each output cell with respect to each input cell, for each function below. In each case the function takes in a tensor $x\in \mathbb{R}^{I}$ and returns a tensor $f(x) \in \mathbb{R}^{O}$, so your job is to compute $\frac{d f(x)_o}{dx_i}$ for all indices $o \in \{0\ldots O-1\}$ and $i\in \{0\ldots I-1\}$. If you get discouraged, just remember, you did this in high school (it just had way less indexing).

### Problem 2: Cosine (0.1p)

Warmup: $f(x_0) = [\cos(x_0)]$

````python
def fb_cos(x):
    f = lambda o: math.cos(x[o])
    dx = lambda i, o: -math.sin(x[o])  # Fill in this line
    return f, dx


in_out(fb_cos, overlap=False, y=gy[1:2], out_shape=1)
````

### Problem 3: Mean (0.2p)

$f(x_0, x_1, \ldots, x_{I-1}) = [(x_0 + x_1 + \ldots + x_{I-1}) / I]$

````python
def fb_mean(x):
    I = x.shape[0]
    f = lambda o: sum(x[i] for i in range(I)) / I
    dx = lambda i, o: 1 / I
    return f, dx


in_out(fb_mean, overlap=False, out_shape=1)
````

### Problem 4: Product (0.2p)

$f(x_0, x_1, \ldots, x_{I-1}) = x_0  x_1  \ldots  x_{I-1}$

````python
def fb_prod(x):
    pr = torch.prod(x)
    f = lambda o: pr
    dx = lambda i, o: pr / x[i]
    return f, dx


in_out(fb_prod, overlap=False, out_shape=1)
````

### Problem 5: Repeat (0.2p)

$f(x_0) = [x_0, x_0,  x_0, \ldots  x_0]$

````python
def fb_repeat(x):
    f = lambda o: x[0]
    dx = lambda i, o: 1
    return f, dx


in_out(fb_repeat, overlap=False, y=gy[1:2], out_shape=50)
````

## Problem 8: ReLU (0.2p)

$$f(x_0, x_1, \ldots) = [\text{relu}(x_0), \text{relu}(x_1), \ldots]$$

Recall

$$
\text{relu}(x) = \begin{cases}
0 & x < 0 \\
x & x >= 0
\end{cases}
$$

(Note: you can ignore the not of non-differentiability at 0.)

````python
def fb_relu(x):
    f = lambda o: (x[o] > 0) * x[o]
    dx = lambda i, o: (i == o)*(x[o] > 0) # Fill in this line
    return f, dx


in_out(fb_relu)
````

## Problem 9: Cumsum (0.3p)

$$f(x_0, x_1, \ldots) = [\sum^0_{i=0} x_{i}, \sum^1_{i=0} x_{i}, \sum^2_{i=0} x_{i}, \ldots, ] / 20$$



````python
def fb_cumsum(x):
    out = torch.cumsum(x, 0)
    f = lambda o: out[o] / 20
    dx = lambda i, o: (i <= o) * 1/20 # Fill in this line
    return f, dx


in_out(fb_cumsum, [20, 35, 40], overlap=True, diff=20)
````

## Problem 10: Running Mean (0.3p)


$$f(x_0, x_1, \ldots)_o = \frac{\displaystyle \sum^o_{i=\max(o-W, 0)} x_i}{ W}$$

````python
def fb_running(x):
    W = 10
    f = lambda o: sum([x[o - do] for do in range(W) if o - do >= 0]) / W
    dx = lambda i, o: (0 <= o - i < W) * 1 / W # Fill in this line
    return f, dx


in_out(fb_running, [0, 20, 35], diff=4)
````

## Problem 11: Sort (0.2p)


$$f(x_0, x_1, \ldots) = \text{x's in sorted order}$$

(This one is a bit counterintuitive! Note that we are not asking you to differentiate the sorting function it self.)

````python
def fb_sort(x):
    sort, argsort = torch.sort(x, 0)
    f = lambda o: sort[o]
    dx = lambda i, o: (i == argsort[o]) # Fill in this line
    return f, dx


in_out(fb_sort, overlap=False)
````

````python
def fb_emean(x, y):
    f = lambda o: (x[o] + y[o]) / 2
    dx = lambda i, o: (i == o) / 2  # Fill in this line
    dy = lambda j, o: (j == o) / 2  # Fill in this line
    return f, dx, dy


zip(fb_emean)
````

## Problem 14: 1D Convolution (0.5p)

This is the standard 1D convolution used in deep learning. There is no wrap-around.

$$f(x, y)_o = \sum_{j=1}^K x_{o+j} * y_{j} / K $$

Note: This is probably the hardest one. The answer is short but tricky.

````python
def fb_conv(x, y):
    W = 5
    f = lambda o: sum((x[o + j] * y[j]) / W for j in range(W))
    # dx = lambda i, o: (0 <= i - o < W) * y[i - o] / W # Fill in this line
    dx = lambda i, o: sum(y[j] / W * (i == o + j) for j in range(W))
    # dy = lambda j, o: x[o + j] / W  # Fill in this line
    dy = lambda j, o: sum(x[o + k] / W * (j == k) for k in range(W))
    return f, dx, dy


SHOW_KERNEL = False
zip(
    fb_conv,
    split=45,
    out_shape=39,
    pos1=[10, 20, 30],
    pos2=[5] if SHOW_KERNEL else [],
    diff=5,
    overlap=True,
)
````

## Problem 15: View (0.3p)

Compute the identity function for all $o,p$. $y$ is ignored.

$$f(X, y)_{o, p} = X_{o, p}$$

````python
def fb_view(x, y):
    f = lambda o, p: x[o, p]
    dx = lambda i, j, o, p: (i == o) * (j == p)  # Fill in this line
    dy = lambda j, o, p: 0  # Fill in this line
    return f, dx, dy


zip(
    make_mat(fb_view, (4, 10), (4, 10)),
    split=40,
    out_shape=40,
    gaps=[10 * i for i in range(4)],
)
````

## Problem 16: Transpose (0.3p)

Transpose row and columns

$$f(X, y)_{o, p} = X_{p, o}$$


````python
def fb_trans(x, y):
    f = lambda o, p: x[p, o]
    dx = lambda i, j, o, p: (i == p) * (j == o)  # Fill in this line
    dy = lambda j, o, p: 0  # Fill in this line
    return f, dx, dy


zip(
    make_mat(fb_trans, in_shape=(4, 10), out_shape=(10, 4)),
    split=40,
    out_shape=40,
    gaps=[10 * i for i in range(4)],
)
````

## Problem 17: Broadcast (0.3p)

Broadcast a matrix with a vector

$$f(X, y)_{o, p} = X_{o, p} \cdot y_p$$

````python
def fb_broad(x, y):
    f = lambda o, p: x[o, p] * y[p]
    dx = lambda i, j, o, p: (i == o) * (j == p) * y[p] # Fill in this line
    dy = lambda j, o, p: x[o,p] * (j == p)  # Fill in this line
    return f, dx, dy


zip(
    make_mat(fb_broad, in_shape=(4, 10), out_shape=(4, 10)),
    split=40,
    out_shape=40,
    gaps=[10 * i for i in range(4)],
)
````

## Problem 18: Mean Reduce (0.3p)

Compute the mean over rows

$$f(X, y)_{o, p} = \sum_{i} X_{i, p} / R$$


````python
def fb_mean(x, y):
    R = x.shape[0]
    f = lambda o, p: sum(x[di, p] for di in range(R)) / R
    dx = lambda i, j, o, p: (j == p) / R# Fill in this line
    dy = lambda j, o, p: 0  # Fill in this line
    return f, dx, dy


zip(
    make_mat(fb_mean, in_shape=(4, 10), out_shape=(1, 10)),
    split=40,
    out_shape=10,
    gaps=[10 * i for i in range(4)],
)
````

## Problem 19: Matmul (0.3p)

Standard matrix multiplication

$$f(X, Y)_{o,p} = \sum_j X_{o, j} Y_{j,p}$$


````python
def fb_mm(x, y):
    _, M = x.shape
    f = lambda o, p: sum(x[o, d] * y[d, p] for d in range(M))
    dx = lambda i, j, o, p: (i == o) * y[j,p]
    dy = lambda i, j, o, p: (j == p) * x[o,i]
    return f, dx, dy


zip(
    make_mat2(fb_mm, in_shape=(5, 5), in_shape2=(5, 5), out_shape=(5, 5)),
    split=25,
    out_shape=25,
    gaps=[5 * i for i in range(5)],
    pos1=None,
    pos2=None,
)
````

## Problem 20: 2D Conv (0.5p)

Two Dimensional convolution

$$f(X, Y)_{o,p} = \sum_{dh}\sum_{dw} X_{o+dh, p+dw} Y_{dh,dw}$$



````python
def fb_conv(x, y):
    kh, kw = y.shape
    f = lambda o, p: sum(
        x[o + di, p + dj] * y[di, dj] / (kh * kw)
        for di in range(kh)
        for dj in range(kw)
    )
    dx = lambda i, j, o, p: (
        y[i - o, j - p] / (kh * kw)
        if(0 <= i - o < kh and 0 <= j - p < kw) # um nicht out of bounds zu fliegen wegen Array zugriff
        else 0 # wegen keinem Zugriff auf das Pixel
    )# Fill in this line
    dy = lambda i, j, o, p: (
        x[i + o, j + p] / (kh * kw) # einfach abschreiben von Formel
    ) # Fill in this line
    return f, dx, dy


zip(
    make_mat2(fb_conv, in_shape=(7, 7), in_shape2=(3, 3), out_shape=(5, 5)),
    split=49,
    out_shape=25,
    gaps=[7 * i for i in range(7)],
    y=torch.cat(2 * [gy.abs() + 0.2], 0)[:58],
    pos1=[23, 36],
    pos2=[],
)
````
